```python
# install missing packages in the notebook kernel
%pip install pandas matplotlib --quiet

import pandas as pd
import matplotlib.pyplot as plt
from IPython.display import display

# Complete project: create a small dataset, analyze, visualize and save results

# dataset (includes "shanto" from the existing cell)
data = {
    "Name": ["shanto", "Alice", "Bob", "Charlie"],
    "Age": [28, 25, 30, 35],
    "Gender": ["M", "F", "M", "M"],
    "Score": [88, 92, 75, 85]
}
df = pd.DataFrame(data)

# display dataframe and basic summaries
print("Dataframe:")
display(df)

print("\nSummary statistics (numeric columns):")
display(df.describe())

print("\nGender distribution:")
display(df["Gender"].value_counts())

# simple visualizations
fig, axes = plt.subplots(1, 2, figsize=(10, 4))
df.plot(x="Name", y="Age", kind="bar", ax=axes[0], legend=False, title="Age by Name")
df.plot(x="Name", y="Score", kind="bar", ax=axes[1], legend=False, title="Score by Name", color="orange")
plt.tight_layout()
plt.show()

# save to CSV
output_path = "people.csv"
df.to_csv(output_path, index=False)
print(f"Saved dataframe to {output_path}")
```
